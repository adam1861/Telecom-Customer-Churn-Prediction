```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LinearRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score , precision_score, recall_score, f1_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
x_train = pd.read_csv("C:\\Users\\adaml\\Documents\\AI_ML_DL\\Projects\\1 - Telecom Customer Churn Prediction\\ready_data\\x_train.csv")
y_train = pd.read_csv("C:\\Users\\adaml\\Documents\\AI_ML_DL\\Projects\\1 - Telecom Customer Churn Prediction\\ready_data\\y_train.csv")
x_test = pd.read_csv("C:\\Users\\adaml\\Documents\\AI_ML_DL\\Projects\\1 - Telecom Customer Churn Prediction\\ready_data\\x_test.csv")
y_test = pd.read_csv("C:\\Users\\adaml\\Documents\\AI_ML_DL\\Projects\\1 - Telecom Customer Churn Prediction\\ready_data\\y_test.csv")
```

```python
#Logistic Regression
logistic_model = LogisticRegression()
logistic_model.fit(x_train, y_train)
y_logistic_pred = logistic_model.predict(x_test)


#Linear Regression
linear_model = LinearRegression()
linear_model.fit(x_train, y_train)
y_linear_pred = linear_model.predict(x_test)
y_linear_pred = (y_linear_pred > 0.5).astype(int)


#Decision Tree
model_decision_tree = DecisionTreeClassifier()
model_decision_tree.fit(x_train, y_train)
y_decision_tree_pred = model_decision_tree.predict(x_test)


#Random Forest
model_random_forest = RandomForestClassifier()
model_random_forest.fit(x_train, y_train)
y_random_forest_pred = model_random_forest.predict(x_test)


#XGBoost
model_xgboost = XGBClassifier()
model_xgboost.fit(x_train, y_train)
y_xgboost_pred = model_xgboost.predict(x_test)
```

```python
#Logistic Regression
print("Logistic Regression - F1 Score:", f1_score(y_test, y_logistic_pred))
print("Logistic Regression - Accuracy:", accuracy_score(y_test, y_logistic_pred))
print("Logistic Regression - Precision:", precision_score(y_test, y_logistic_pred))
print("Logistic Regression - Recall:", recall_score(y_test, y_logistic_pred))
print("\n")
 

#Linear Regression
print("Linear Regression - F1 Score:", f1_score(y_test, y_linear_pred))
print("Linear Regression - Accuracy:", accuracy_score(y_test, y_linear_pred))
print("Linear Regression - Precision:", precision_score(y_test, y_linear_pred))
print("Linear Regression - Recall:", recall_score(y_test, y_linear_pred))
print("\n")


#Decision Tree
print("Decision Tree - F1 Score:", f1_score(y_test, y_decision_tree_pred))
print("Decision Tree - Accuracy:", accuracy_score(y_test, y_decision_tree_pred))
print("Decision Tree - Precision:", precision_score(y_test, y_decision_tree_pred))
print("Decision Tree - Recall:", recall_score(y_test, y_decision_tree_pred))
print("\n")


#Random Forest
print("Random Forest - F1 Score:", f1_score(y_test, y_random_forest_pred))
print("Random Forest - Accuracy:", accuracy_score(y_test, y_random_forest_pred))
print("Random Forest - Precision:", precision_score(y_test, y_random_forest_pred))
print("Random Forest - Recall:", recall_score(y_test, y_random_forest_pred))
print("\n")


#XGBoost
print("XGBoost - F1 Score:", f1_score(y_test, y_xgboost_pred))
print("XGBoost - Accuracy:", accuracy_score(y_test, y_xgboost_pred))
print("XGBoost - Precision:", precision_score(y_test, y_xgboost_pred))
print("XGBoost - Recall:", recall_score(y_test, y_xgboost_pred))
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

y_train_nn = y_train.values.ravel()
y_test_nn = y_test.values.ravel()
```

```python
x_train_nn, x_val_nn, y_train_nn_split, y_val_nn_split = train_test_split(
    x_train_scaled,
    y_train_nn,
    test_size=0.2,
    random_state=42,
    stratify=y_train_nn
)

negatives = np.sum(y_train_nn_split == 0)
positives = np.sum(y_train_nn_split == 1)
total = len(y_train_nn_split)
class_weight = {
    0: total / (2.0 * negatives),
    1: total / (2.0 * positives)
}

nn_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(x_train_scaled.shape[1],)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Dropout(0.25),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

nn_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=8,
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=4,
    min_lr=1e-5
)

history = nn_model.fit(
    x_train_nn,
    y_train_nn_split,
    validation_data=(x_val_nn, y_val_nn_split),
    epochs=50,
    batch_size=32,
    class_weight=class_weight,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)
```

```python
val_prob = nn_model.predict(x_val_nn).ravel()
thresholds = np.arange(0.20, 0.61, 0.01)
val_f1_scores = [f1_score(y_val_nn_split, (val_prob >= threshold).astype(int)) for threshold in thresholds]
best_threshold = thresholds[int(np.argmax(val_f1_scores))]

y_nn_prob = nn_model.predict(x_test_scaled).ravel()
y_nn_pred = (y_nn_prob >= best_threshold).astype(int)

print("Best validation threshold:", best_threshold)
print("Neural Network - F1 Score:", f1_score(y_test_nn, y_nn_pred))
print("Neural Network - Accuracy:", accuracy_score(y_test_nn, y_nn_pred))
print("Neural Network - Precision:", precision_score(y_test_nn, y_nn_pred))
print("Neural Network - Recall:", recall_score(y_test_nn, y_nn_pred))
```

```python
summary_table = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Linear Regression',
        'Decision Tree',
        'Random Forest',
        'XGBoost',
        'Neural Network'
    ],
    'Accuracy': [
        accuracy_score(y_test, y_logistic_pred),
        accuracy_score(y_test, y_linear_pred),
        accuracy_score(y_test, y_decision_tree_pred),
        accuracy_score(y_test, y_random_forest_pred),
        accuracy_score(y_test, y_xgboost_pred),
        accuracy_score(y_test_nn, y_nn_pred)
    ],
    'Precision': [
        precision_score(y_test, y_logistic_pred),
        precision_score(y_test, y_linear_pred),
        precision_score(y_test, y_decision_tree_pred),
        precision_score(y_test, y_random_forest_pred),
        precision_score(y_test, y_xgboost_pred),
        precision_score(y_test_nn, y_nn_pred)
    ],
    'Recall': [
        recall_score(y_test, y_logistic_pred),
        recall_score(y_test, y_linear_pred),
        recall_score(y_test, y_decision_tree_pred),
        recall_score(y_test, y_random_forest_pred),
        recall_score(y_test, y_xgboost_pred),
        recall_score(y_test_nn, y_nn_pred)
    ],
    'F1 Score': [
        f1_score(y_test, y_logistic_pred),
        f1_score(y_test, y_linear_pred),
        f1_score(y_test, y_decision_tree_pred),
        f1_score(y_test, y_random_forest_pred),
        f1_score(y_test, y_xgboost_pred),
        f1_score(y_test_nn, y_nn_pred)
    ]
})

summary_table = summary_table.sort_values(by='F1 Score', ascending=False).reset_index(drop=True)
summary_table
```
